```python
# Dataset path
dataset_path = os.path.join(
    "..",
    "dataset",
    "student_data.csv"
)

# Load CSV file
df = pd.read_csv(dataset_path)

print("Dataset Loaded Successfully!")

df.head()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Student Marks Correlation")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="StudyHours",
    y="FinalMarks"
)

plt.title("Study Hours vs Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="FinalMarks"
)

plt.title("Attendance vs Final Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="PreviousMarks",
    y="FinalMarks"
)

plt.title("Previous Marks vs Final Marks")
plt.xlabel("Previous Marks")
plt.ylabel("Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="SleepHours",
    y="FinalMarks"
)

plt.title("Sleep Hours vs Final Marks")
plt.xlabel("Sleep Hours")
plt.ylabel("Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="SocialMediaHours",
    y="FinalMarks"
)

plt.title("Social Media Hours vs Final Marks")
plt.xlabel("Social Media Hours")
plt.ylabel("Final Marks")

plt.show()
```

```python
correlation_matrix = df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance Correlation Heatmap")

plt.show()
```

```python
features = [
    "StudyHours",
    "Attendance",
    "PreviousMarks",
    "SleepHours",
    "SocialMediaHours"
]

for feature in features:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df,
        x=feature,
        kde=True
    )

    plt.title(f"Distribution of {feature}")

    plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df)

plt.title("Outlier Analysis")

plt.xticks(rotation=45)

plt.show()
```

```python
average_marks = df["FinalMarks"].mean()

print("Average Final Marks:", round(average_marks, 2))
```

```python
highest_marks = df["FinalMarks"].max()

lowest_marks = df["FinalMarks"].min()

print("Highest Marks:", round(highest_marks, 2))

print("Lowest Marks:", round(lowest_marks, 2))
```

```python
correlation_with_target = df.corr()["FinalMarks"]

correlation_with_target = correlation_with_target.sort_values(
    ascending=False
)

print(correlation_with_target)
```

```python
correlation_with_target = df.corr()["FinalMarks"]

correlation_with_target = correlation_with_target.sort_values(
    ascending=False
)

print(correlation_with_target)
```

```python
correlation_with_target = df.corr()["FinalMarks"]

correlation_with_target = correlation_with_target.sort_values(
    ascending=False
)

print(correlation_with_target)
```

```python
print("========== EDA SUMMARY ==========")

print("Total Students:", df.shape[0])

print("Total Features:", df.shape[1] - 1)

print("Target Variable: FinalMarks")

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Rows:", df.duplicated().sum())

print("Average Final Marks:", round(df["FinalMarks"].mean(), 2))

print("Maximum Final Marks:", round(df["FinalMarks"].max(), 2))

print("Minimum Final Marks:", round(df["FinalMarks"].min(), 2))

print("=================================")
```
